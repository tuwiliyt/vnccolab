# 🍎 macOS Hack di Google Colab (Docker `tuwiliyt/macoshack`)

Notebook ini menjalankan image Docker **[tuwiliyt/macoshack](https://hub.docker.com/r/tuwiliyt/macoshack)** langsung di Google Colab dengan akses GUI berbasis web via **Cloudflare Tunnel**.

### 🌟 Fitur Utama:
- 📦 **Docker Container Ready**: Otomatis pull dan jalankan `tuwiliyt/macoshack:latest`.
- 📁 **Integrasi Google Drive**: Otomatis me-mount Google Drive ke dalam folder `/config/GoogleDrive` container.
- 🌐 **Web GUI Langsung**: Akses antarmuka grafis langsung melalui tautan Cloudflare di browser (PC/HP).

```python
#@title 🚀 Jalankan macOS Hack Container
#@markdown Centang untuk menghubungkan Google Drive Anda:
Hubungkan_Google_Drive = True #@param {type:"boolean"}

if Hubungkan_Google_Drive:
    from google.colab import drive
    drive.mount('/content/drive')

!curl -sL https://raw.githubusercontent.com/tuwiliyt/vnccolab/main/macoshack.sh | bash
```
